## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Escribe tu código aquí
taxis['hora'] = taxis['pickup'].dt.hour
taxis['dia_num'] = taxis['pickup'].dt.dayofweek
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()
taxis['fin_semana'] = taxis['dia_num'].isin([5,6])
taxis['fecha'] = taxis['pickup'].dt.date
taxis['cruza_medianoche'] = taxis['fecha'] != taxis['fecha'].shift(1)

taxis.head()
```

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Escribe tu código aquí
print(f"Número total de viajes antes de la limpieza: {taxis.shape[0]}")

# Condición 1: Viajes con pickup antes de marzo
excluded_feb_mask = taxis['pickup'] < '2019-03-01'
excluded_feb_count = excluded_feb_mask.sum()
print(f"Viajes excluidos con 'pickup' antes de marzo de 2019: {excluded_feb_count}")

# Condición 2: Viajes con velocidad promedio mayor a 100 mph
#Primero, asegurarnos de no incluir NaNs en el conteo de exclusiones. Luego, filtrar por velocidad > 100 mph
excluded_speed_mask = (taxis['velocidad_mph'] > 100) & (taxis['velocidad_mph'].notna())
excluded_speed_count = excluded_speed_mask.sum()
print(f"Viajes excluidos con 'velocidad_mph' > 100 mph: {excluded_speed_count}")

# Combinar las condiciones de exclusión para crear el dataset limpio
# La máscara para incluir viajes será lo contrario de las máscaras de exclusión
include_mask = (~excluded_feb_mask) & (~excluded_speed_mask)
taxis_limpio = taxis[include_mask].copy()

total_excluded_count = taxis.shape[0] - taxis_limpio.shape[0]
print(f"Número total de viajes excluidos: {total_excluded_count}")
print(f"Número total de viajes en 'taxis_limpio': {taxis_limpio.shape[0]}")

# Verificación de que taxis_limpio cumple las condiciones:
# No debería haber ningún viaje con pickup antes de 2019-03-01
print(f"Verificación: Viajes con 'pickup' < 2019-03-01 en taxis_limpio: {(taxis_limpio['pickup'] < '2019-03-01').sum()}")
# No debería haber ningún viaje con velocidad_mph > 100
print(f"Verificación: Viajes con 'velocidad_mph' > 100 en taxis_limpio: {(taxis_limpio['velocidad_mph'] > 100).sum()}")
```

```python
# Escribe tu código aquí
# Filtrar viajes que inician entre las 7:00 y 9:59
hora_inicio = 7
hora_fin = 9

periodo_manana_mask = (taxis_limpio['hora'] >= hora_inicio) & (taxis_limpio['hora'] <= hora_fin)
viajes_manana = taxis_limpio[periodo_manana_mask]

# Separar en días hábiles y fines de semana
dias_habiles_manana = viajes_manana[viajes_manana['fin_semana'] == False]
fines_semana_manana = viajes_manana[viajes_manana['fin_semana'] == True]

# Reportar métricas para días hábiles
print("Viajes en días hábiles (7:00-9:59)\n")
print(f"Número de viajes: {dias_habiles_manana.shape[0]}")
print(f"Mediana de duración (min): {dias_habiles_manana['duracion_min'].median():.2f}")
print(f"Mediana de velocidad (mph): {dias_habiles_manana['velocidad_mph'].median():.2f}")

# Reportar métricas para fines de semana
print("\nViajes en fines de semana (7:00-9:59)\n")
print(f"Número de viajes: {fines_semana_manana.shape[0]}")
print(f"Mediana de duración (min): {fines_semana_manana['duracion_min'].median():.2f}")
print(f"Mediana de velocidad (mph): {fines_semana_manana['velocidad_mph'].median():.2f}")
```
