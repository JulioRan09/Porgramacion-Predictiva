- El **porcentaje** de filas sobrantes respecto al total.

```python
# Escribe tu código aquí
#Si contaramos con conservar un dato de los duplicados (ya sea con first o last), podemos calcular el porcentaje de filas sobrantes como
#el total de filas menos las filas de duplicados / total de filas

total_filas = len(hoteles)
filas_sobrantes = total_filas - hoteles_first
porcentaje_filas_sobrantes = (filas_sobrantes / total_filas) * 100
print(f'El porcentaje de filas sobrantes respecto al total es: {porcentaje_filas_sobrantes:.2f}%')
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# Escribe tu código aquí
# Número de filas antes de eliminar duplicados
filas_antes = hoteles.shape[0]
print(f"Número de filas antes de eliminar duplicados: {filas_antes}")

# Crear hoteles_limpio eliminando los duplicados exactos, conservando la primera aparición
hoteles_limpio = hoteles.drop_duplicates(keep='first')

# Renumerar el índice
hoteles_limpio = hoteles_limpio.reset_index(drop=True)

# Número de filas después de eliminar duplicados
filas_despues = hoteles_limpio.shape[0]
print(f"Número de filas después de eliminar duplicados: {filas_despues}")

# Cuántas filas se eliminaron
filas_eliminadas = filas_antes - filas_despues
print(f"Número de filas eliminadas: {filas_eliminadas}")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Escribe tu código aquí
# Para el DataFrame original (hoteles)
resumen_original = hoteles.groupby('hotel').agg(
    num_reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    adr_promedio=('adr', 'mean')
).reset_index()

print("Resumen de métricas para el DataFrame original:\n")
display(resumen_original)

print("\n" + "==============================" + "\n")

# Para el DataFrame sin duplicados (hoteles_limpio)
resumen_limpio = hoteles_limpio.groupby('hotel').agg(
    num_reservas=('hotel', 'size'),
    tasa_cancelacion=('is_canceled', 'mean'),
    adr_promedio=('adr', 'mean')
).reset_index()

print("Resumen de métricas para el DataFrame limpio (sin duplicados):\n")
display(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Escribe tu código aquí
# Agregar la columna 'version' a cada DataFrame resumen
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

# Unir los DataFrames con pd.concat()
comparacion = pd.concat([resumen_original, resumen_limpio])

# Renumerar el índice y ordenar por 'hotel'
comparacion = comparacion.reset_index(drop=True).sort_values(by='hotel')

print("Tabla comparativa de métricas por hotel:\n")
display(comparacion)
```

### 4.2 Duplicados parciales (10 pts)

La columna `reservation_status_date` registra la fecha del último cambio de estado. Detecta las filas de `hoteles_limpio` que **coinciden en todas las columnas excepto `reservation_status_date`**.

1. Cuenta cuántas filas sobrantes hay con ese criterio.

Después responde: ¿qué podrían representar estas filas? ¿Las eliminarías? Justifica.

```python
# Escribe tu código aquí
# Columnas a considerar para detectar duplicados parciales (todas excepto 'reservation_status_date')
columnas_para_parciales = [col for col in hoteles_limpio.columns if col != 'reservation_status_date']

# 1. Contar cuántas filas sobrantes hay con ese criterio (duplicados parciales)
# Usamos keep=False para identificar todas las instancias de los duplicados parciales
duplicados_parciales = hoteles_limpio[hoteles_limpio.duplicated(subset=columnas_para_parciales, keep=False)]
num_filas_parciales_sobrantes = duplicados_parciales.duplicated(subset=columnas_para_parciales, keep='first').sum()
print(f"Número de filas duplicadas parciales (idénticas excepto reservation_status_date): {num_filas_parciales_sobrantes}")

# 2. Mostrar algunas de esas filas ordenadas, con las columnas especificadas
columnas_a_mostrar = ['hotel', 'lead_time', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month', 'adr', 'reservation_status', 'reservation_status_date']

# Ordenar por las columnas de identificación para que los grupos parciales aparezcan juntos
duplicados_parciales_ordenados = duplicados_parciales.sort_values(by=columnas_para_parciales)

display(duplicados_parciales_ordenados[columnas_a_mostrar].head(10))

# 3. Mostrar la distribución de reservation_status en esas filas
print("\nDistribución de 'reservation_status' en las filas duplicadas parciales:")
display(duplicados_parciales['reservation_status'].value_counts(normalize=True))
```
